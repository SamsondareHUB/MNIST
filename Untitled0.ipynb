```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import numpy as np

# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Load MNIST dataset
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))
])

train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root='./data', train=False, download=True, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=1000, shuffle=False)

print(f"Training samples: {len(train_dataset)}")
print(f"Test samples: {len(test_dataset)}")
```

```python
# Define the Neural Network
class MNISTNet(nn.Module):
    def __init__(self):
        super(MNISTNet, self).__init__()
        self.conv1 = nn.Conv2d(1, 32, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.pool = nn.MaxPool2d(2, 2)
        self.fc1 = nn.Linear(64 * 7 * 7, 128)
        self.fc2 = nn.Linear(128, 10)
        self.dropout = nn.Dropout(0.25)

    def forward(self, x):
        x = self.pool(torch.relu(self.conv1(x)))
        x = self.pool(torch.relu(self.conv2(x)))
        x = x.view(-1, 64 * 7 * 7)
        x = torch.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        return x

# Create model instance
model = MNISTNet().to(device)
print("Model created successfully!")
print(f"Total parameters: {sum(p.numel() for p in model.parameters())}")
```

```python
# Training configuration
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training function
def train(model, device, train_loader, optimizer, epoch):
    model.train()
    total_loss = 0
    correct = 0
    total = 0

    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(device), target.to(device)
        optimizer.zero_grad()
        output = model(data)
        loss = criterion(output, target)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        pred = output.argmax(dim=1, keepdim=True)
        correct += pred.eq(target.view_as(pred)).sum().item()
        total += target.size(0)

        if batch_idx % 200 == 0:
            print(f'Epoch {epoch} [{batch_idx * len(data)}/{len(train_loader.dataset)}] '
                  f'Loss: {loss.item():.4f} Accuracy: {100. * correct / total:.2f}%')

    return total_loss / len(train_loader), 100. * correct / total

# Test function
def test(model, device, test_loader):
    model.eval()
    test_loss = 0
    correct = 0

    with torch.no_grad():
        for data, target in test_loader:
            data, target = data.to(device), target.to(device)
            output = model(data)
            test_loss += criterion(output, target).item()
            pred = output.argmax(dim=1, keepdim=True)
            correct += pred.eq(target.view_as(pred)).sum().item()

    test_loss /= len(test_loader)
    accuracy = 100. * correct / len(test_loader.dataset)
    print(f'\nTest set: Average loss: {test_loss:.4f}, Accuracy: {correct}/{len(test_loader.dataset)} ({accuracy:.2f}%)\n')
    return test_loss, accuracy

# Train for 3 epochs
print("Starting training...")
for epoch in range(1, 4):
    train_loss, train_acc = train(model, device, train_loader, optimizer, epoch)
    test_loss, test_acc = test(model, device, test_loader)

print("Training completed!")
```

```python
# Visualize some predictions
import random

def visualize_predictions(model, test_loader, num_images=10):
    model.eval()
    images_shown = 0

    fig, axes = plt.subplots(2, 5, figsize=(15, 6))
    axes = axes.ravel()

    with torch.no_grad():
        for data, target in test_loader:
            data, target = data.to(device), target.to(device)
            output = model(data)
            pred = output.argmax(dim=1)

            for i in range(len(data)):
                if images_shown >= num_images:
                    break

                img = data[i].cpu().squeeze()
                true_label = target[i].item()
                pred_label = pred[i].item()

                axes[images_shown].imshow(img, cmap='gray')
                color = 'green' if true_label == pred_label else 'red'
                axes[images_shown].set_title(f'True: {true_label}, Pred: {pred_label}', color=color)
                axes[images_shown].axis('off')
                images_shown += 1

            if images_shown >= num_images:
                break

    plt.tight_layout()
    plt.show()

print("Visualizing predictions (green = correct, red = incorrect):")
visualize_predictions(model, test_loader)
```

```python
# Confusion Matrix and Accuracy per Digit
from sklearn.metrics import confusion_matrix
import seaborn as sns

def evaluate_detailed(model, test_loader):
    model.eval()
    all_preds = []
    all_targets = []

    with torch.no_grad():
        for data, target in test_loader:
            data, target = data.to(device), target.to(device)
            output = model(data)
            pred = output.argmax(dim=1)
            all_preds.extend(pred.cpu().numpy())
            all_targets.extend(target.cpu().numpy())

    # Confusion Matrix
    cm = confusion_matrix(all_targets, all_preds)

    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=True)
    plt.title('Confusion Matrix - MNIST Digit Classification')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.show()

    # Per-digit accuracy
    print("\nAccuracy per digit:")
    for digit in range(10):
        digit_mask = np.array(all_targets) == digit
        digit_acc = np.sum((np.array(all_preds)[digit_mask] == digit)) / np.sum(digit_mask) * 100
        print(f"Digit {digit}: {digit_acc:.2f}%")

print("Generating confusion matrix and per-digit accuracy...")
evaluate_detailed(model, test_loader)
```

```python
# Show some misclassified examples (if any)
def show_errors(model, test_loader, num_errors=5):
    model.eval()
    errors_found = 0

    fig, axes = plt.subplots(1, min(num_errors, 5), figsize=(15, 3))
    if num_errors == 1:
        axes = [axes]

    with torch.no_grad():
        for data, target in test_loader:
            data, target = data.to(device), target.to(device)
            output = model(data)
            pred = output.argmax(dim=1)

            # Find incorrect predictions
            incorrect = pred != target

            for i in range(len(data)):
                if incorrect[i] and errors_found < num_errors:
                    img = data[i].cpu().squeeze()
                    true_label = target[i].item()
                    pred_label = pred[i].item()
                    confidence = torch.softmax(output[i], dim=0)[pred_label].item() * 100

                    axes[errors_found].imshow(img, cmap='gray')
                    axes[errors_found].set_title(f'True: {true_label}\nPred: {pred_label} ({confidence:.1f}%)',
                                                 color='red', fontsize=10)
                    axes[errors_found].axis('off')
                    errors_found += 1

                    if errors_found >= num_errors:
                        break

            if errors_found >= num_errors:
                break

    if errors_found == 0:
        print("No errors found in the first batch!")
    else:
        plt.tight_layout()
        plt.show()
        print(f"\nFound {errors_found} misclassified examples")

print("Looking for misclassified examples:")
show_errors(model, test_loader, num_errors=5)
```
